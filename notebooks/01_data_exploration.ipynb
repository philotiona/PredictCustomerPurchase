# Data Exploration

In this notebook, we will explore the dataset of customer purchases. We will visualize data distributions and understand relationships between variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = pd.read_csv('../data/raw/large_purchases.csv')

# Display the first few rows of the dataset
data.head()
```

```python
# Summary statistics
data.describe()
```

```python
# Visualize the distribution of purchases by item
plt.figure(figsize=(12, 6))
sns.countplot(data=data, x='item_id')
plt.title('Distribution of Purchases by Item')
plt.xlabel('Item ID')
plt.ylabel('Number of Purchases')
plt.xticks(rotation=45)
plt.show()
```

```python
# Visualize purchases over time
data['date'] = pd.to_datetime(data['date'])
purchases_over_time = data.groupby(data['date'].dt.date).size()

plt.figure(figsize=(12, 6))
purchases_over_time.plot(kind='line')
plt.title('Purchases Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Purchases')
plt.grid()
plt.show()
```
